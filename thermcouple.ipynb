```python
def find_char_boxes(thresh_img, min_x=150):
    contours, _ = cv2.findContours(thresh_img, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    boxes = [cv2.boundingRect(c) for c in contours]

    keep = []
    for b in boxes:
        x, y, w, h = b
        is_digit = h > 50
        if is_digit and x >= min_x:     # only accept digit-shaped blobs to the RIGHT of min_x
            keep.append(b)

    keep.sort(key=lambda b: b[0])
    return keep

boxes1 = find_char_boxes(thresh1)
boxes2 = find_char_boxes(thresh2)
print("Box1 chars:", boxes1)
print("Box2 chars:", boxes2)
```

```python
segments = {
    "a": (0.2, 0.8, 0.0, 0.15),
    "b": (0.7, 1.0, 0.1, 0.5),
    "c": (0.7, 1.0, 0.5, 0.9),
    "d": (0.2, 0.8, 0.85, 1.0),
    "e": (0.0, 0.3, 0.5, 0.9),
    "f": (0.0, 0.3, 0.1, 0.5),
    "g": (0.2, 0.8, 0.4, 0.6),
}

def fraction_on(region):
    if region.size == 0:
        return 0
    return (region == 255).sum() / region.size

def read_digit(digit_crop, threshold=0.4):
    h, w = digit_crop.shape
    lit = set()
    for name, (fx0, fx1, fy0, fy1) in segments.items():
        region = digit_crop[int(fy0*h):int(fy1*h), int(fx0*w):int(fx1*w)]
        if fraction_on(region) > threshold:
            lit.add(name)
    return lit
```

```python
PATTERNS = {
    frozenset("abcdef"): "0",
    frozenset("bc"): "1",
    frozenset("abged"): "2",
    frozenset("abgcd"): "3",
    frozenset("fgbc"): "4",
    frozenset("afgcd"): "5",
    frozenset("afgedc"): "6",
    frozenset("abc"): "7",
    frozenset("abcdefg"): "8",
    frozenset("abcdfg"): "9",
}

def lit_to_char(lit):
    return PATTERNS.get(frozenset(lit), "?")
```

```python
DECIMALS = 1   # how many digits appear after the decimal point on this display

chars = ""
for (x, y, w, h) in boxes1:
    if w > 30 and h < 25:        # minus sign
        chars += "-"
        continue
    crop = thresh1[y:y+h, x:x+w]
    lit = read_digit(crop)
    chars += lit_to_char(lit)

# insert the decimal point before the last DECIMALS digits
sign = ""
if chars.startswith("-"):
    sign = "-"
    chars = chars[1:]

value_str = sign + chars[:-DECIMALS] + "." + chars[-DECIMALS:]
value = float(value_str)
print("Reading:", value)
```

```python
def read_display(thresh_img, decimals=1):
    boxes = find_char_boxes(thresh_img)
    chars = ""
    for (x, y, w, h) in boxes:
        if w > 30 and h < 25:        # minus sign
            chars += "-"
            continue
        crop = thresh_img[y:y+h, x:x+w]
        lit = read_digit(crop)
        chars += lit_to_char(lit)

    # reject frames where a "-" showed up somewhere it shouldn't
    if chars.count("-") > 1:
        return None
    if "-" in chars[1:]:   # a dash anywhere except the very first character
        return None

    sign = ""
    if chars.startswith("-"):
        sign = "-"
        chars = chars[1:]

    if not chars or "?" in chars:
        return None

    try:
        value_str = sign + chars[:-decimals] + "." + chars[-decimals:]
        return float(value_str)
    except ValueError:
        return None
```

```python
def to_thresh(crop):
    gray = cv2.cvtColor(crop, cv2.COLOR_BGR2GRAY)
    _, thresh = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)

    kernel = np.ones((3, 3), np.uint8)
    thresh = cv2.morphologyEx(thresh, cv2.MORPH_CLOSE, kernel)

    return thresh
```
